Question_Answer LLM Model

Loading Data

```python
from langchain_community.document_loaders import TextLoader
```

```python
loader= TextLoader(file_path= "C:\\Users\\Priya Adhwaria\\Desktop\\LLM Project\\QA LLM\\train-v2.0.json")
```

```python
data= loader.load()
#data
```

Chunks Creation

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
text_splitter= RecursiveCharacterTextSplitter(chunk_size=20000, chunk_overlap=10, separators=["\n", "]"])
```

```python
split_data= text_splitter.split_documents(data)
len(split_data)
```

Data Embedding

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings= HuggingFaceEmbeddings()
```

Vector Database

```python
from langchain_community.vectorstores import FAISS
```

```python
vectordb= FAISS.from_documents(documents= split_data, embedding= embeddings)
```

Prompt Engineering

```python
from langchain_core.prompts import PromptTemplate
```

```python
temp= """ For given following context and query, provide answer of query from the context of most relevant chunk obtained from retriever from the source document only. If the answer is not present in the context, show 'I don't know'. Don't hallucinate the answer.
summaries: {summaries}
question: {question}"""

prompt= PromptTemplate(template= temp, input_variables= ["chunk", "summaries"])
```

Initiating LLM Model

```python
import os
import dotenv
```

```python
dotenv.load_dotenv("QA_API_Key.env")
```

```python
key= os.getenv("my_api_key")
```

```python
from langchain_google_genai import GoogleGenerativeAI
```

```python
llm_model= GoogleGenerativeAI(model= "gemini-2.5-flash", api_key= key)
```

```python
#quest= llm_model.invoke("What is LLM Model?")
#print(quest)
```

RAG/ Retrieval Chain

```python
from langchain_classic.chains import RetrievalQAWithSourcesChain
```

```python
retriever= vectordb.as_retriever()
```

```python
chain= RetrievalQAWithSourcesChain.from_llm(llm= llm_model, retriever= retriever, combine_prompt= prompt)
```

```python
chain.invoke("What is LLM Model?")
```

```python
chain.invoke("Beyonce is married to whom?")
```
